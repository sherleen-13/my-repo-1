```python
import os

def list_files(startpath):
    for root, dirs, files in os.walk(startpath):
        level = root.replace(startpath, '').count(os.sep)
        indent = ' ' * 4 * (level)
        print(f'{indent}{os.path.basename(root)}/')
        subindent = ' ' * 4 * (level + 1)
        for f in files[:5]:  # Limit output to first 5 files per directory
            print(f'{subindent}{f}')
        if len(files) > 5:
            print(f'{subindent}... ({len(files) - 5} more files)')

target_drive_folder = '/content/drive/MyDrive/IAT360'
if os.path.exists(target_drive_folder):
    print(f"Structure of {target_drive_folder}:")
    list_files(target_drive_folder)
else:
    print(f"{target_drive_folder} does not exist. Listing /content/drive/MyDrive instead:")
    list_files('/content/drive/MyDrive')
```

```python
import os
from ultralytics import YOLO

# 1. Target directory on your mounted Google Drive (added the space in 'IAT 360')
drive_path = '/content/drive/MyDrive/IAT 360/CV-TestData-Sherleen'
os.makedirs(drive_path, exist_ok=True)

# 2. Write data.yaml directly inside your Google Drive dataset folder
yaml_content = f'''path: {drive_path}
train: train/images
val: train/images

nc: 7
names: ['plastic', 'biodegradable', 'metal', 'glass', 'paper', 'cardboard', 'generaltrash']'''

yaml_path = os.path.join(drive_path, 'data.yaml')
with open(yaml_path, 'w') as f:
    f.write(yaml_content)

print(f"Successfully created data.yaml at: {yaml_path}")

# 3. Load and train the model using your data.yaml
model = YOLO('yolov8n.pt')
trainingResults = model.train(
    data=yaml_path,
    epochs=3
)

# 4. Validate model performance
validationResults = model.val()

# 5. Predict on one of your training images to test it
testingResults = model.predict(
    source=os.path.join(drive_path, 'train/images/'),
    save=True
)
```
